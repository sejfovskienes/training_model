```python
import os
os.environ["USE_TF"] = "0"
```

```python
import pandas as pd
from transformers import pipeline
```

**Preprocessing**

```python
path = "../gdelt_gold_news_dataset.csv"

df = pd.read_csv(path)

df.head()
```

```python
df["date"] = pd.to_datetime(df["date"], format="%Y%m%d")
df = df.sort_values("date")
```

```python
df.head()
```

```python
df.tail()
```

**Sentiment Analysis**

```python
classifier = pipeline(
    "text-classification",
    model="mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis",
    tokenizer="mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis"
)
```

```python
df["sentiment"] = df["text"].apply(lambda x: classifier(x)[0]['label'])
```

```python
df.head()
```

```python
path = "../gdelt_gold_news_sentiments.csv"
df.to_csv(path, index=False)
```
