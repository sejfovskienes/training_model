```python
import os 
import pandas as pd
from pydantic_ai import Agent
from dotenv import load_dotenv
from pydantic_ai.models.google import GoogleModel
from pydantic_ai.providers.google import GoogleProvider
```

```python
load_dotenv(override=True)
API_KEY = os.environ['GEMMA_API_KEY']
```

```python
SYSTEM_PROMPT = """
    You are a helpful assistant that converts the url link of a news article to sentence like topic. Make sure not to over explain.
    Try to approximate the topic with same number of words as in the link. Answer only the topic, do not add extra discussion. 
    Make sure that you capture all the necessary information and the details about the url link.
"""
```

```python
provider = GoogleProvider(api_key=API_KEY)
model = GoogleModel("gemini-2.0-flash", provider=provider)
agent = Agent(model=model, system_prompt=SYSTEM_PROMPT, output_type=str)
```

```python
#--- idea: run the agent for every link in the dataset.
```

```python
query = "https://elkodaily.com/news/local/business/mining/students-learn-min/"
answer = await agent.run(query)
print(answer.output)
```

```python
path = r"C:\Users\User\dev\training_model\gdelt_data_fetch\gdelt_gold_filtered\20140101_gold_filtered.csv"
df = pd.read_csv(path)
```

```python
url_bulk = ""

for url in df['SOURCEURL']:
    url_bulk = url_bulk + f" {url}"

print(url_bulk)
```

```python
SYSTEM_PROMPT = """
    You are a helpful assistant that converts the url link of a news article to sentence like topic. As an input you will get bulk of links as string.
    Your task is to return two or maximum three sentences that are going to give some type of description about what happened to gold market or gold prices.
    Try to approximate the situation as close as you can. Answer only the topic, do not add extra discussion. 
    Make sure that you capture all the necessary information and the details about the text that is provided you.
"""

provider = GoogleProvider(api_key=API_KEY)
model = GoogleModel("gemini-2.0-flash", provider=provider)
agent = Agent(model=model, system_prompt=SYSTEM_PROMPT, output_type=str)

answer = await agent.run(url_bulk)
print(answer.output)
```
