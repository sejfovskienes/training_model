In this notebook we will try to train the model using news and making sentiment of them using data source from gdeltproject.org.

**Fetching news**

```python
import requests
import zipfile
import io
import datetime
import os
import time

# Directory to save extracted CSVs
SAVE_DIR = "gdelt_data"
os.makedirs(SAVE_DIR, exist_ok=True)

def daterange(start_date, end_date):
    for n in range(int((end_date - start_date).days) + 1):
        yield start_date + datetime.timedelta(n)

def download_gdelt_event(date):

    date_str = date.strftime("%Y%m%d")
    url = f"http://data.gdeltproject.org/events/{date_str}.export.CSV.zip"
    print(f"Downloading {url} ...")

    try:
        response = requests.get(url, timeout=30)
        response.raise_for_status()  

        
        with zipfile.ZipFile(io.BytesIO(response.content)) as z:
           
            for filename in z.namelist():
                print(f"Extracting {filename} ...")
                # Save CSV file
                csv_path = os.path.join(SAVE_DIR, filename)
                with open(csv_path, 'wb') as f_out:
                    f_out.write(z.read(filename))
                print(f"Saved to {csv_path}")
        return True
    except requests.HTTPError as e:
        print(f"HTTP error: {e} (Probably no data for {date_str})")
    except Exception as e:
        print(f"Error downloading or extracting for {date_str}: {e}")
    return False
```

```python
start_date = datetime.date(2014, 1, 1)
end_date = datetime.date.today()

for single_date in daterange(start_date, end_date):
    success = download_gdelt_event(single_date)
    if not success:
        print(f"Skipping date {single_date}")
    time.sleep(1)  
```
